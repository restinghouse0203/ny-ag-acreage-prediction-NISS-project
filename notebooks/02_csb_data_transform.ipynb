```python
import geopandas as gpd

# Filter New York (STATEFIPS == '36')
ny_csb = csb_data[csb_data['STATEFIPS'] == '36'].copy()

# 1. Parquet (compressed, good for big data)
ny_csb.to_parquet(INTERIM_DIR / 'ny_csb_20172024' / 'ny_csb.parquet_20172024', engine='pyarrow')

# 2. Feather (fastest, keeps geometry)
ny_csb.to_feather(INTERIM_DIR / 'ny_csb_20172024' / 'ny_csb.feather_20172024')

# 3. CSV (simple, no geometry preserved)
ny_csb.drop(columns='geometry').to_csv(INTERIM_DIR / 'ny_csb_20172024' / 'ny_csb.csv_20172024', index=False)
```
